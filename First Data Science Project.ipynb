##### Step 4: Visualise the Data

- Show distribution of key variables
- Compare different groups
- Explore relationships between variables


```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
plt.hist(df["JAMB_Score"],bins=20)
plt.title("Distribution of JAMB Scores")
plt.xlabel(" JAMB Score")
plt.ylabel("Number of students")
plt.show()      
```

The distribution of Jamb score shows how scores are spread across students.Most students scored within the middle range,while fewer students achieved very low or high score


```python
sns.boxplot(x="Study_Hours_Per_Week",
y="JAMB_Score", data=df) 
plt.title("Study Hours and JAMB Scores")
plt.show()           
```

Student Who studied for more hours per week generally recorded higher JAMB scores compared to those who studied less

```python
sns.scatterplot(
    x="Attendance_Rate",
    y="JAMB_Score",
    data=df
)

plt.title("Attendance Rate vs JAMB Score")
plt.show()
```

There is a positive relationship between attendance rate and JAMB score. Students with higher attendance rates tend to perform better in the examination

```python
numeric_df = df.select_dtypes(include=["int64","float64"])

plt.figure(figsize=(10,6))
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Matrix")
plt.show()
```

**Observations**:

The heatmap helps identify which variables are most strongly related to JAMB scores. Variables with stronger positive correlations are likely to have a greater influence on student performance
